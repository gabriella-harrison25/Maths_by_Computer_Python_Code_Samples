#### (illustrating idea of rates of change with respect to the model parameters and gradient descent)

Consider a really small dataset consisting of only three points in the plane: $(1,1),(2,4),(3,5)$.

For each point (training example) consider the $x$-coordinate as a single input feature, and the $y$-coordinate as the output or label of the example. The objective is to find the best line to fit these data. Here 'best' will mean the line which minimises the mean of the sum of the squares of the residuals with respect to the three points.

In this task and in the following tasks A3 and A4 we will use a slightly different convention to store our data with respect to the one used in A1 or the lectures. $X_{\text{train}}$ and $y_{\text{train}}$ will now be the transpose of the corresponding matrices in A1. The feature matrix $X_{\text{train}}$ contains the feature vectors for all the training examples. More precisely $X_{\text{train}}= \begin{pmatrix} 1 & 2 & 3 \end{pmatrix} $ has one column for each training example and each column is the feature vector for that training example (in this case we have only one feature which is the $x$-coordinate).

The labels matrix $y_{\text{train}}$ contains the labels for all the training examples. More precisely $y_{\text{train}} = \begin{pmatrix} 1 & 4 & 5\end{pmatrix}$ has one column for each training example and each column contains the label for that training example (the $y$-coordinate).

The model here is a simple linear model

$$
\hat y = mx +c
$$

with trainable parameters $m$ and $c$. The goal is to find the optimal values of $m$ and $c$ to fit these training data. In the Machine Learning context this means defining a loss function between prediction and true label:

$$
L(\hat y, y)=(\hat y - y)^2
$$

and a cost function which is simply the average on all training examples of these losses.

This means that the value of $m$ is updated to $m - \displaystyle\frac{\partial J}{\partial m}\alpha$ and the value of $c$ is updated to $c - \displaystyle\frac{\partial J}{\partial c}\alpha$ in each iteration.

**1)** **Loss function for (1,1):** \
       \begin{equation*} \begin{split}L(m(1)+c,1) &= L(m+c, 1)\\
        &= (m+c-1)^2\\
        &= m^2 + 2mc - 2m +c^2 - 2c + 1 \end{split} \end{equation*} \
   **Loss function for (2,4):** \
       \begin{equation*}\begin{split}
       L(m(2)+c,4) &= L(2m+c, 4)\\
       & = (2m+c-4)^2 \\
       &= 4m^2 + 4mc - 16m +c^2 - 8c + 16 \end{split}\end{equation*} \
   **Loss function for (3,5):**\
       \begin{equation*}\begin{split} L(m(3)+c,5) &= L(3m+c, 5)\\
       &= (3m+c-5)^2\\
       &= 9m^2 + 6mc - 30m +c^2 - 10c + 25 \end{split}\end{equation*}

   The cost function is the mean of these 3 losses:\
   $$J=\frac{L(m+c, 1) + L(2m+c, 4) + L(3m+c, 5)}{3} = \frac{m^2 + 2mc - 2m +c^2 - 2c + 1 + 4m^2 + 4mc - 16m +c^2 - 8c + 16 + 9m^2 + 6mc - 30m +c^2 - 10c + 25}{3} $$
   Simplifying by collecting terms, we get:\
   $$ J = \frac{14m^2 + 12mc - 48m + 3c^2 - 20c + 42}{3} $$
   as required.


**2)** The partial derivatives are as follows:\
 $\frac{\partial J}{\partial m} = \frac{28m + 12c - 48}{3}$ \
   $\frac{\partial J}{\partial c} = \frac{12m + 6c - 20}{3}$


```python
# Part 3
from numpy import random #Needed to generate random m and c between -2 and 2

def partial_derivatives(m_, c): # Calculates the partial derivatives using formulas found in part 2.
  Jm = (28*m_+12*c-48)/3
  Jc = (12*m_+6*c-20)/3
  return(Jm, Jc)

def cost(m_,c): # Calculates the cost function using expression from part 1.
  J = (14*(m_**2) + 12*m_*c -48*m_ + 3*(c**2) - 20*c +42)/3
  return(J)

def model1(alpha, num_iterations):
  m_ = random.uniform(-2,2)
  c = random.uniform(-2,2)
  for i in range(0, num_iterations):
    partial_m, partial_c = partial_derivatives(m_,c)
    if i%(num_iterations//5) == 0:
      print(cost(m_,c)) # Returns cost function 5 times during the iterations.
    m_ = m_ - partial_m*alpha
    c = c - partial_c*alpha
  return(m_,c)

print("The final values for m and c are:", model1(0.1, 1000))
```

```python
  # Part 4
import numpy as np
import matplotlib.pyplot as plt

# Plot optimal line
x = np.linspace(-5,10,301)
m_,c = model1(0.01,1000)
y = m_*x + c
plt.plot(x,y)

#Plot datapoints
x_train = np.array([1,2,3])
y_TRAIN = np.array([1,4,5])
plt.plot(x_train, y_TRAIN, '*') #use different data name than the one in A1
plt.xlabel("X-Values")
plt.ylabel("Y-Values")
plt.title("Linear Regression from Scratch Model")
plt.legend(["Model Solution", "Datapoints"])
plt.show()
```

#### Initialize parameters

**Complete the function `initialize_parameters` started below.** It takes in the dimension of the input layer `n_X` and the dimension of the output layer `n_y` and returns returns a Python dictionary `parameters`. The dictionary `parameters` contains the initialized parameters of the model: the weight matrix `W` and the bias vector `b`.

Note that `n_X = 2` in our example, each point has two input coordinates,  and the dimension of the output layer is `n_y = 1`. This means that `W` will be a $1\times 2$ matrix and `b` will be a $1\times 1$ matrix for our example.
```python
def initialize_parameters(n_X, n_y):
    """
    Argument:
    n_X -- integer
    n_y -- integer

    Returns:
    parameters -- python dictionary containing your parameters:
                    W -- weight matrix
                    b -- bias vector
    """

    #Initialize weight matrix nd bias vector
    W = np.random.randn(n_y, n_X) * 0.01
    b = np.zeros((n_y, 1))

    #Store W and b in a dictionary parameters
    parameters = {"W": W,
                  "b": b}

    return parameters
```

#### Forward Propagation

Next you will implement forward propagation for logistic regression:

$$
z = WX + b \text{ and then }
\hat y = \sigma(z)
$$

As above, for $\lambda\in\mathbb{R}$, $\sigma(\lambda)=\displaystyle\frac{1}{1+\exp(-\lambda)}$ (this is applied entry-wise to $z$ in the above).

**Start by implementing the sigmoid function in numpy by replacing *None* in the code cell below with appropriate code. Then run the cell.**

```python
def sigmoid(z):
    """
    Argument:
    z -- input data, numpy array

    Returns:
    numpy array that contains sigmoid function applied to each
    element of z.
    """
    return 1/(1+np.exp(-z))
```

**Complete the function `forward_propagation` started below.** It calculates the output of the logistic regression model, called `y_hat`, and returns (as a tuple) `y_hat` and `z` (in the notation given above).

**Once you have completed the code run this cell.**

```python
def forward_propagation(X_flower, parameters):
    """
    Argument:
    X -- input data, numpy array of dimension (2, 400). Every column of X
         is the feature vector for a single training example (input of the logistic regression)
    parameters -- python dictionary containing the values of W and b to be used
                  to compute the forward propagation.

    Returns:
    y_hat -- numpy array of dimension (1,400). Every column of y_hat is the
             output probability of the logistic regression for that training example.
    z -- numpy array of dimension (1,400), result of linear transformation before sigmoid activation is applied.
    """

    # retrieve W and b from the dictionary "parameters"

    W, b=parameters["W"], parameters["b"]

    # compute z
    z=np.dot(W,X_flower) + b  #in this case np.dot works, but typically we use W*X for scalar multiplication of matrices (Gabriella)

    # Compute y_hat
    y_hat = sigmoid(z)

    return y_hat, z
```

#### Compute cost

Given the predictions $\hat y^{(i)}$ on all the training examples, you can compute the cost $J$ as the average over all training examples of the losses:
$$
J = \frac{1}{m} \sum\limits_{i = 0}^{m} L(\hat y^{(i)}, y^{(i)})  .
$$
Here we are using the binary cross-entropy loss:
$$
L(\hat y^{(i)}, y^{(i)}) = -\large\left(\small y^{(i)}\log\left(\hat y^{(i)}\right) + (1-y^{(i)})\log\left(1- \hat y^{(i)}\right)  \large  \right).
$$

Recall $y^{(i)}$ is the $i^{\text{th}}$ component in $y$, i.e. the label for example $i$ and $\hat{y}^{(i)}$ is the models current prediction of this label, the $i^{\text{th}}$ component in $\hat{y}$.

**Complete the function `compute_cost()`, started below, to compute the value of the cost $J$:**

1. First compute the vector of losses element-wise using the numpy arrays given as inputs `y_hat` and `y`.

2. Sum the losses and divide by the number of training examples `m`. Call this value `cost`.

3. Cast `cost` as type `float` using `float()` and return it.

**Once you have completed the code run this cell.**

```python
def compute_cost(y_hat, y_flower):
    """
    Arguments:
    y_hat -- numpy array, output of forward propagation
    y -- numpy array of the same shape as y_hat, containing the true labels (0 for red, 1 for blue)

    Returns:
    cost -- a float which is the cross-entropy cost
    """

    # retrieve number of training examples from the shape of y
    m = y_flower.shape[1]

    # compute the vector of losses by computing the cross-entropy loss element-wise
    L = - (y_flower*np.log(y_hat) + (1-y_flower)*np.log(1-y_hat))

    # compute the total cost by averaging the loss over all training examples
    cost2 = np.sum(L) / m

    # cast cost as a float
    cost2=float(cost2)

    return cost2
```

#### Backpropagation

You can now implement backward propagation. Note that you will complete the code cell below this, much of the code is already given to you but you need to stick with the notation established.

NOTATION: in general for a generic numpy array `M` and a function $F$ with the entries of `M` as inputs, denote by `dM` the numpy array representing the gradient of $F$ with respect to `M`. More precisely `dM` contains the partial derivatives of $F$ with respect to the entries of `M`. For example if
$$M = \begin{pmatrix} a_{0,0} & a_{0,1} & a_{0,2} \\ a_{1,0} & a_{1,1} &  a_{1,2}\end{pmatrix}$$ and $F$ is a function of $a_{0,0},a_{0,1},a_{0,2},a_{1,0},a_{1,1},a_{1,2}$ then $$dM = \begin{pmatrix} \frac{\partial F}{\partial a_{0,0}} & \frac{\partial F}{\partial a_{0,1}} & \frac{\partial F}{\partial a_{0,2}} \\ \frac{\partial F}{\partial a_{1,0}} & \frac{\partial F}{\partial a_{1,1}} & \frac{\partial F}{\partial a_{1,2}}\end{pmatrix}.$$

Throught this project $F$ will be the cost function $J$.

**Complete the function `backward_propagation()` started below.** It takes in a dictionary `parameters` containing the current parameters, `y_hat`,`X` and `y` and it returns a dictionary containing the gradients `dW` and `db` of the cost function with respect to the trainable parameters `W` and `b`.

You are given the key code for calculating `dW` and `db`, it is, in the notation already established.

`dz = y_hat - y
dW = np.dot(dz, X.T)/m
db = np.sum(dz, axis=1, keepdims=True)/m`

You will see these lines of code in the cell below.

**Once you have completed the code run this cell.**


```python
def backward_propagation(parameters, y_hat, X_flower, y_flower):
    """
    Implement the backward propagation.

    Arguments:
    parameters -- python dictionary containing W and b to use
    y_hat - from output of forward prop
    X -- input data
    y -- labels vector

    Returns:
    grads -- python dictionary containing the gradients
    """
    # Retrieve the number of training examples from the shape of y
    m2 = y_flower.shape[1]

    # retrieve W from the dictionary "parameters".
    W = parameters["W"]

    # Backward propagation: calculate dW, and db.
    dz = y_hat - y_flower
    dW = np.dot(dz, X_flower.T)/m2
    db = np.sum(dz, axis=1, keepdims=True)/m2

    grads = {"dW": dW,
             "db": db}

    return grads
```

#### Update parameters

Next you will define a function to update parameters using gradient descent. In a similar way to that seen in task A2,

1. $W$ will be updated to

$$W - dW\times \text{learning rate}.$$

and

2. $b$ will be updated to$$b - db\times\text{learning rate}.$$

**Complete the function `update_parameters()` started below.** It takes in the dictionary `parameters` containing the current parameters `W` and `b`, the dictionary `grads` containing the current gradient `dW`,`dB` and the learning rate (which defaults to the value $1.2$). **Once you have completed the code run this cell.**

```python
def update_parameters(parameters, grads, learning_rate = 1.2):
    """
    Arguments:
    parameters -- python dictionary containing the parameters
    grads -- python dictionary containing the gradients

    Returns:
    parameters -- python dictionary containing the updated parameters
    """
    # Retrieve a copy of each parameter from the dictionary "parameters". Use copy.deepcopy(...) for W

    W = copy.deepcopy(parameters["W"])
    b = parameters["b"]

    # Retrieve each gradient from the dictionary "grads"
    dW, db=grads["dW"], grads["db"]

    # Update each parameter
    W=W - dW*learning_rate
    b=b - db*learning_rate

    # Store the new parameters in the dictionary parameters
    parameters["W"]= W
    parameters["b"] = b

    return parameters
```

#### Model

Now you will put everything together into a function `model` where you will apply `update_parameters` iteratively to update the parameters a number of times.

**Complete the function `model` started below.** It takes as inputs the training data `X`, `y`, a positive integer `num_iterations` which is the number of times the parameters will be updated (defaulting to $10,000$) and a boolean `print_cost` which, when set to `True`, will mean the function prints out some costs as the values of the parameters are updated). `model` return a dictionary containing the optimised parameters at the end of this process. **Once you have completed the code run this cell.**

```python
def model(X_flower, y_flower, num_iterations2 = 10000, print_cost=False):
    """
    Arguments:
    X -- numpy array, input data
    y -- numpy array, true labels
    num_iterations -- number of iterations in the gradient descent loop
    print_cost -- if True, print the cost every 1000 iterations

    Returns:
    parameters -- python dictionary which contains the optimized parameters
                  learnt by the model.
    """
    # Optional: to control the random seed of the initialization uncomment next line:
    # np.random.seed(3)

    # retrieve n_x and n_y from X and y
    n_X = X_flower.shape[0]
    n_y = y_flower.shape[0]

    # Initialize parameters
    parameters=initialize_parameters(n_X, n_y)

    # Training loop (gradient descent)
    for i in range(0, num_iterations2):

        # Forward propagation
        y_hat, z= forward_propagation(X_flower, parameters)

        # Compute the cost
        cost2=compute_cost(y_hat, y_flower)

        # Backpropagation
        grads= backward_propagation(parameters, y_hat, X_flower, y_flower)

        # Update parameters (use learning_rate = 1.2)
        parameters=update_parameters(parameters, grads, 1.2)

        # Print the cost every 1000 iterations
        if print_cost and i % 1000 == 0:
            print ("Cost after iteration %i: %f" %(i, cost2))

    learned_parameters = parameters

    # return learned parameters
    return learned_parameters
```

#### Predict on new data

Below you are given a complete function `predict` that uses the learned parameters (output of `model`) and the `forward_propagation` function to predict a class for each example in the input matrix `X`. **Run the code cell below.**

```python
def predict(learned_parameters, X_flower):
    """
    Arguments:
    parameters -- python dictionary containing the learned parameters given by model
    X -- numpy array input data

    Returns
    predictions -- vector of predictions for each column of X (red: 0 / blue: 1)
    """

    # Compute predicted probabilities using the learned parameters

    y_hat, z = forward_propagation(X_flower, learned_parameters)

    # Classify as 0 or 1 using 0.5 as a threshold

    predictions = (y_hat > 0.5)

    return predictions
```

```python
def model_nn_2(X_flower, y_flower, n_h = 4, num_iterations3 = 10000, learning_rate2 = 1.2):
    """
    Arguments:
    X -- numpy array, input data
    y -- numpy array, true labels
    n_h -- integer, number of neurons in the hidden layer
    num_iterations -- number of iterations in the gradient descent loop
    print_cost -- if True, print the cost every 1000 iterations

    Returns:
    parameters -- optimized parameters learnt by the model.
    """
    # Optional: to control the random seed of the initialization uncomment next line
    # np.random.seed(3)

    # retrieve n_x and n_y
    n_x2 = X_flower.shape[0]
    n_y2 = y_flower.shape[0]

    # Initialize parameters
    W2 = np.random.randn(n_h, n_x2) * 0.01
    b2 = np.zeros((n_y2, 1))

    T2 = np.random.randn(n_h, n_x2).reshape(n_x2, n_h) * 0.01
    c2 = np.zeros((n_y2, 1))

    #make new params dictionary
    parameters_2 = {"W2":W2, "b2":b2}

    # Training loop
    for i in range(0, num_iterations3):

        # Forward propagation
        z2 = np.dot(W2, X_flower) + b2
        a2 = np.tanh(z2)
        v2 = np.dot(T2, a2) + c2
        y_hat2 = sigmoid(v2)

        # Compute cost
        cost_2=compute_cost(y_hat2, y_flower) ##FLAG - NEED TO REDEF COMPUTE COST?

        # Backpropagation
        m_2 = y_flower.shape[1]
        dv_2 = y_hat2 - y_flower

        # First two variables to learn
        dT_2 = np.dot(dv_2, a2.T)/m_2
        dc_2 = np.sum(dv_2, axis=1, keepdims=True)/m_2

        dz_2 = np.dot(T2.T, dv_2) * (1 - np.power(a2, 2))

        # Two more variables to learn
        dW_2 = np.dot(dz_2, X_flower.T)/m_2
        db_2 = np.sum(dz_2, axis=1, keepdims=True)/m_2

        # Update parameters
        W2 = W2 - dW_2*learning_rate2
        b2 = b2 - db_2*learning_rate2
        T2 = T2 - dT_2*learning_rate2
        c2 = c2 - dc_2*learning_rate2

    learned_parameters_2 = {"W2": W2, "b2": b2, "T2": T2, "c2": c2}

    return learned_parameters_2
```

```python
def predict_nn_2(learned_parameters_2, X_flower):
    """
    Returns
    predictions -- vector of predictions for each column of X (red: 0 / blue: 1)
    """

    W2, b2, T2, c2 = learned_parameters_2["W2"], learned_parameters_2["b2"], learned_parameters_2["T2"], learned_parameters_2["c2"]
    z2 = np.dot(W2, X_flower) + b2
    a2 = np.tanh(z2)
    v2 = np.dot(T2, a2) + c2
    y_hat2 = sigmoid(v2)

    predictions_2 = (y_hat2 > 0.5)

    return predictions_2
```

$$\textbf{How can we further improve neural networks?}$$

<br>

When training a neural network there are two main ways it can go wrong: underfitting and overfitting [7]. Understanding both problems, why they happen, and how to fix them is one of the most important ideas in machine learning.

$$\textbf{Overfitting}$$

<br>

$$\textbf{What is it?}$$

Overfitting is when a model learns the training data too well [8]. It picks up not just the underlying pattern but also the noise and random quirks specific to the training set. The result is a model that scores very highly on training data but performs much worse on new unseen data [7,8].

A good analogy is a student who memorises every past exam paper word for word rather than understanding the topic. They would ace a repeated past paper but struggle badly on a new one [8].

<br>

$$\textbf{Why does it happen?}$$

- The model has too many parameters i.e. too many neurons or layers relative to the amount of training data [8]
- The training dataset is too small
- The model is trained for too many iterations and starts memorising rather than learning

<br>

$$\textbf{Applications Where Overfitting is a Problem}$$

- Medical diagnosis: a model trained on a small hospital's patient records might memorise patterns specific to that hospital and fail on data from elsewhere [7]
- Fraud detection: financial fraud patterns change over time so a model that overfits historical data quickly becomes useless on new transactions [7]

<br>

$$\textbf{Pros and Cons}$$

- Pro: can capture complicated non-linear patterns in data [8]
- Con: much more likely to overfit, especially with limited training data [8]
- Con: harder to interpret and explain what the model has learned [8]

<br>

$$\textbf{Fixes}$$

- Use more training data so the model has more genuine patterns to learn from [7]
- Use a simpler model with fewer neurons or layers [7]
- Use regularisation - adds a penalty to the cost function to discourage large weights (demonstrated later with Lasso and Ridge Regression [2]) [8]

Regularisation is a technique used to prevent overfitting. It does this by adding penalty terms and encouraging simpler models. There are 3 types:


*   Lasso Regularisation

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification


x_gen, y_gen = make_classification(n_samples=1000, n_features=100, n_informative = 40, n_repeated = 5, n_redundant=5, n_classes=2)
xgen_train, xgen_test, ygen_train, ygen_test = train_test_split(x_gen, y_gen, test_size=0.3, random_state=42)
lr = LinearRegression().fit(xgen_train, ygen_train)

print("The R2 value for the train data is:")
print(lr.score(xgen_train, ygen_train))
print("The R2 value for the test data is:")
print(lr.score(xgen_test, ygen_test))
```

### Lasso Regression

Lasso stands for Least Absolute Shrinkage and Selection Operator, it uses a L1 regularisation technique [2,5]. It adds the absolute value of the coefficient to the loss function. This means some coefficients can shrink to zero, meaning only important features are selected.

$$ Cost = \frac{1}{n} \sum_{i=1}^{n}(y_i - \hat y_i)^2 + \alpha\sum_{i=1}^m |w_i|$$

```python
#This code utilized source [5].
from sklearn.linear_model import Lasso
# Alpha = 1
lasso_model_1 = Lasso(alpha = 1).fit(xgen_train, ygen_train)
print("\n When alpha = 1, number of non-zero features:", sum(lasso_model_1.coef_ != 0))
print("The R2 value for the train data is:")
print(lasso_model_1.score(xgen_train, ygen_train))
print("The R2 value for the test data is:")
print(lasso_model_1.score(xgen_test, ygen_test))

# Alpha = 0.5
lasso_model_2 = Lasso(alpha = 0.5).fit(xgen_train, ygen_train)
print("\n When alpha = 0.5, the  number of non-zero features:", sum(lasso_model_2.coef_ != 0))
print("The R2 value for the train data is:")
print(lasso_model_2.score(xgen_train, ygen_train))
print("The R2 value for the test data is:")
print(lasso_model_2.score(xgen_test, ygen_test))

# Alpha = 0.1
lasso_model_3 = Lasso(alpha = 0.1).fit(xgen_train, ygen_train)
print("\n When alpha = 0.1, number of non-zero features:", sum(lasso_model_3.coef_ != 0))
print("The R2 value for the train data is:")
print(lasso_model_3.score(xgen_train, ygen_train))
print("The R2 value for the test data is:")
print(lasso_model_3.score(xgen_test, ygen_test))

# Alpha = 0.01
lasso_model_4 = Lasso(alpha = 0.01).fit(xgen_train, ygen_train)
print("\n When alpha = 0.01, number of non-zero features:", sum(lasso_model_4.coef_ != 0))
print("The R2 value for the train data is:")
print(lasso_model_4.score(xgen_train, ygen_train))
print("The R2 value for the test data is:")
print(lasso_model_4.score(xgen_test, ygen_test))
```

We can see when alpha is relatively big, lasso regression is not a very good model, as the $R^2$ is very small. This is because lasso regression shrinks too many features to zero and means the model now underfits. We can see as alpha decreases, the number of non-zero features (out of 100) increases and so does the $R^2$ of the test data. The $R^2$ for the train data actually decreases, but this is because the model is now less complicated, so will slightly underfit. However, the $R^2$ for the test data is closer to 1, so the model is now better, and overfits less.

### Ridge Regression

Ridge regression models use the L2 regularisation technique [2,3,9]. Ridge Regression adds the squared value of the coefficient to the penalty term. This stops as many features being completely shrunk to zero, instead it shrinks the coefficients of correlated features. This avoids the model underfitting the data.
$$ Cost = \frac{1}{n} \sum_{i=1}^{n}(y_i - \hat y_i)^2 + \alpha\sum_{i=1}^m w_i^2$$

```python
#This code utilizes source [6].
from sklearn.linear_model import Ridge

ridge_model_1 = Ridge(alpha = 0.7).fit(xgen_train, ygen_train)
print("The R2 value for the train data is:")
print(ridge_model_1.score(xgen_train, ygen_train))
print("The R2 value for the test data is:")
print(ridge_model_1.score(xgen_test, ygen_test))
```

We found the ridge regression model didn't make much difference to the R squared value for this dataset. This is because ridge regression tends to work better on datasets with more features. In general, ridge regression does tend to be more effective than lasso regression as it is less likely to underfit the data.

### Elastic Net Regression

Elastic net regression uses a mix of L1 (lasso) and L2 (ridge) regularisation techniques [2,3,4]. We add the absolute value of the coefficients and the square of coefficients. There is an extra parameter which helps control the ratio of L1 and L2 regression.
$$ Cost = \frac{1}{n}  \sum_{i=1}^{n}(y_i - \hat y_i)^2 + \alpha(
  \lambda\sum_{i=1}^m |w_i |+(1-\lambda)\sum_{i=1}^m w_i^2)$$
In the sklearn ElasticNet function the parameter which controls the ratio of L1 and L2 regression is called l1_ratio. When l1_ratio = 1, the ElasticNet function is the same as the lasso function, and when l1_ratio = 0, ridge regression occurs.

```python
#This code utilizes information from source [4].
from sklearn.linear_model import ElasticNet

Elastic_Net_Model_1 = ElasticNet(alpha = 0.01, l1_ratio = 0.01).fit(xgen_train, ygen_train)
print("The R2 value for the train data is:")
print(Elastic_Net_Model_1.score(xgen_train, ygen_train))
print("The R2 value for the test data is:")
print(Elastic_Net_Model_1.score(xgen_test, ygen_test))
```

How it works [10,11,14]:

This is explained in more detail in section A but here is a brief summary:
* the relationship can be modeled as a linear equation:
$$Y= w_0 + w_1 X + \epsilon $$
where:
  * $Y$ is the dependent variable (target).
  * $X$ is the independent variable (feature).
  * $w_0$ is the y-intercept of the line.
  * $w_1$ is the slope of the line, representing the effect of $X$ on $Y$.
  * $\epsilon$ is the error term, capturing the variability in $Y$ not explained by $X$.
* The main goal of linear regression is to find the best-fit line through a set of data points that minimises the difference between the actual values and predicted values by estimating the parameters $w_0$, $w_1$ etc.
* The steps:
  * Hypothesis − We assume that there is a linear relation between input and output.
  * Cost Function − Define a loss or cost function. The cost function quantifies the model's prediction error. The cost function takes the model's predicted values and actual values and returns a single scaler value that represents the cost of the model's prediction.
  * Optimisation − Optimise (minimise) the model's cost function by updating the model's parameters.

**Polynomial Regression**


 While the linear regression model is able to understand patterns for a given dataset by fitting in a simple linear equation, it might not be accurate when dealing with complex data. In those instances, we need to come up with curves which adjust with the data rather than the lines. One approach is to use a polynomial regression model [15]. The polynomial regression is a special case of multiple linear regression but there is a difference that multiple linear regression assumes linearity of input features [15].

  This is in the form:

$$ y= w_0 + w_1x +w_2 x^2 + w_3 x^3 + ... + w_n x^n + \epsilon$$

  where:
  * $y$ is the dependent variable (output).
  * $x$ is the independent variable (input).
  * $w_0, w_1, w_2,..., w_n$ are the coefficients (parameters) of the model.
  * $n$ is the degree of the polynomial (the highest power of $x$).
  * $\epsilon$ is the error term or residual, representing the difference between the observed value and the model's prediction.

 <br>

 How it works [15]:
 * Polynomial regression works in a similar way as linear regression works. It is modelled as multiple linear regression. The input feature is transformed into polynomial features of higher degrees (${x^2, x^3, ..., x^n}$). These features are now treated as separate independent variables as in multiple linear regression. Now, a multiple linear regressor is trained on these transformed polynomial features.

 <br>

 Applications [10]:
* Agricultural Yield Prediction Based on Environmental Conditions:
  
  Predict the crop yield based on variables such as temperature, rainfall, and soil quality, where the relationship between these factors and yield is not linear.
* Modeling Electricity Consumption in Relation to Temperature:
  
  Forecast the electricity consumption of a city based on the temperature, where consumption increases during extreme cold and hot temperatures but drops at moderate temperatures.

  <br>

Advantages [10]:
* works on any size data set
* works well for non-linear problems

<br>

Disadvantages [10]:
* we need to choose the right polynomial degree for a good bias/variance trade off

 <br>

## Logistic Regression (As seen in section A)

Logistic Regression is a supervised machine learning algorithm used for classification problems [21]. Unlike linear regression which predicts continuous values it predicts the probability that an input belongs to a specific class [22].
* It is used for binary classification where the output can be one of two possible categories such as Yes/No, True/False or 0/1.
* For binary classifiction, it uses sigmoid function ($\sigma(\lambda)=\displaystyle\frac{1}{1+\exp(-\lambda)}$) to convert inputs into a probability value between 0 and 1.

<br>

Types [21]:
* Binary Logistic Regression -  The target (dependent) variable has only two possible outcomes. Examples include Yes/No, Pass/Fail or 0/1. It is the most common form of logistic regression and is used for binary classification problems.

* Multinomial Logistic Regression - The target (dependent) variable has three or more nominal categories that are not ordered.  For example, classifying animals into categories like "cat," "dog" or "sheep." It extends the binary logistic regression to handle multiple classes.

* Ordinal Logistic Regression: This type applies when the dependent variable has three or more categories with a natural order or ranking. Examples include ratings like "low," "medium" and "high." It takes the order of the categories into account when modeling.

<br>

How does it work?
* Again this is explained in more detail in section A but to summarise the idea [21]:

  Logistic regression model transforms the linear regression function continuous value output into categorical value output using a sigmoid function which maps any real-valued set of independent variables input into a value between 0 and 1. This function is known as the logistic function.

<br>

Advantages [24]:
* Because of its efficient and straightforward nature, it doesn't require high computation power, is easy to implement, easily interpretable, and used widely by data analysts and scientists. Also, it doesn't require scaling of features.

Disadvantages [24]:
* Logistic regression is not able to handle a large number of categorical features/variables. It is vulnerable to overfitting. Also, it can't solve the non-linear problems, which is why it requires a transformation of non-linear features. Logistic regression will not perform well with independent variables that are not correlated to the target variable and are very similar or correlated to each other.

For example, if the threshold is 0.5 then a probability greater than or equal to 0.5 means a prediction of class 1, otherwise, class 0.

  When generating the ROC curve, different probability thresholds are tested. For each threshold it calculates the True Positive Rate and the False Positive Rate. This allows the model’s ability to distinguish between the two classes to be assessed across all possible thresholds.
  * lower threshold - more positives predicted
  * higher threshold - fewer positives predicted
* So the curve shows the trade-off between correctly detecting positive cases and incorrectly predicting positives.

**Multinomial Logistic Regression Model**

What is it?
* Modified version of logistic regression that predicts a multinomial probability (i.e. more than two classes) for each input example. It predicts probabilities for each class and selects the most likely outcome [25].

<br>

How does it work? [23, 25]
* Binary logistic regression models the probability of one outcome versus another. Multinomial logistic regression does the same thing, but repeatedly. It picks one category as a reference group, then estimates a separate set of comparisons for every other category against that reference. If your outcome has four categories, the model fits three separate equations, each one comparing a category to the reference. The number of equations is always one fewer than the number of outcome categories.
* The model assigns each category a raw score based on the predictor variables and their associated weights. These raw scores don’t mean much on their own because they can be any number, positive or negative. To convert them into meaningful probabilities, the model uses a function (called the softmax function) that does two things: it forces every score to be positive, and it divides each score by the total of all scores. The result is a set of probabilities, one for each category, that are all between 0 and 1 and add up to 1.
* How the probabilities for each class are calculated - the softmax function (an extension of the sigmoid function see in Binary Logistic Regression).
  * For a vector of scores $z=[z_1,...,z_k]$ where $k$ is the number of classes: $$
\sigma (z_i) = \frac{\exp(z_i)}{\sum\limits_{j = 1}^{k} (\exp(z_j))}
$$
* The next step is to select the class with maximum probability using the argmax function:
  * $\hat y= $ argmax $ \sigma(z)$
* The cost function for a multinomial regression model is also known as a cross entropy function. This function penalises the classification model when it estimates a low probability for the target or actual class.

* For example, if you’re predicting which of three transportation modes a commuter will choose (car, bus, train), the model produces a probability for each option for every individual in your data. A given commuter might get 0.65 for car, 0.25 for bus, and 0.10 for train. The model’s final prediction is whichever category has the highest probability.

<br>

Applications [25]:
* Marketing / Customer Analytics to predict which product a customer will choose. Products are the classes and categories such as age and income are the features.
* Medical diagnosis - predict which disease category a patient belongs to based on symptoms. Diseases are the classes and the symptoms are the features.
* Image Classification - predict which object category appears in an image.

<br>

Binary vs Multinomial Logistic Regression[22,23,25]:

* Number of Outcome Categories:
  * Binary: two outcome categories.
  * Multinomial: more than two outcome categories.
* Model Complexity:
  * Binary: simpler as it involves single categories.
  * Multinomial: more complex than binary as it accounts for the multiple categories.
* Interpretation of Coefficients:
  * Binary: coefficients represent the log odds ratio of the event occurring
  * Multinomial: they compare each category to the reference category.
* Data Structure:
  * Binary: deals with the binary outcomes.
  * Multinomial: requires the outcome variable to be nominal or ordinal.

```python
#This code utilizes data from source [17]

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import numpy as np

# Load the Iris dataset for multi-class classification
iris2 = load_iris()

# For understanding purposes we will condider two feature petal length, petal width. This means we can plot a decision aboundary because if we used all 4 features, we cannot visualise a decision boundary in 4-D.
X_nom = iris2.data[:, (2, 3)]
y_nom = iris2.target

# Split the data into training and testing sets
X_trainNom, X_testNom, y_trainNom, y_testNom = train_test_split(X_nom, y_nom, test_size=0.2, random_state=26)

# Standardise the features
scaler = StandardScaler()
X_trainNom = scaler.fit_transform(X_trainNom)
X_testNom = scaler.transform(X_testNom)

# Create and train the softmax regression model (no need to specifiy multinomial because Logisticregression detects that automatically)
softmax_model = LogisticRegression(solver='lbfgs', random_state=42)

softmax_model.fit(X_trainNom, y_trainNom)

# Make predictions on the test set
y_predNom = softmax_model.predict(X_testNom)

# Evaluate the softmax regression model
accuracy = accuracy_score(y_testNom, y_predNom)
print(f"Accuracy: {accuracy * 100:.2f}%")

print()

# Display classification report
print("Classification Report:")
print(classification_report(y_testNom, y_predNom))

# Predict the class of a new observation (petal length = 2, petal width = 3)
softmax_model.predict([[2, 3]])

# Predict the class probabilities for a new observation
print("class 0:", softmax_model.predict_proba([[2, 3]])[0][0])
print("class 1:", softmax_model.predict_proba([[2, 3]])[0][1])
print("class 2:", softmax_model.predict_proba([[2, 3]])[0][2])
```

This all seems expected because the Iris data set is fairly easy to separate using petal measurements.

## K-Nearest Neighbors

*This section utilizes information from sources [27,28,29]*

##### Background Information

$\textbf{What is K-Nearest Neighbors?}$

K-Nearest Neighbors is a **supervised, instance-based** learning method used for classification in machine learning models [28]. It only performs the computations required at the time of classification. A K-Nearest Neighbor model utilizes an input value of *K* to determine the number of nearest data points taken into account when making a decision [27].

<br>

$\textbf{Why is it useful?}$

* Recommendation Systems [27]
  * Programs that suggest similar movies or products use K-Nearest Neighbor algorithms by comparing users' preferences to those of others.

* Customer Segmentation [27]
  * Customers can be easily grouped based on shopping behaviours. This can improve market research for particular products.

* Speech Recognition [28]
  * K-Nearest Neighbors models can match spoken words to known patterns to identify words.

* Economics/Credit Evaluation [28]
  * These algorithms can also be used to assess credit-worthiness for a loan applicant.

<br>

$\textbf{What are the challenges?}$

* K-Nearest Neighbor models can be **slow with large data sets** because they calculate the distance to every point. However, other data structures, like Ball-Tree, minimize computational inefficiencies [28].

* The accuracy of K-Nearest Neighbor models drops when there are too many features in a dataset. This is known as the **"Curse of Dimensionality."** [28]

* These models can also **overfit data** when it is unclean or highly-dimensional [27]. This is another side effect of the Curse of Dimensionality. To reduce the impact of overfitting, feature selection and dimension reduction techniques can be used.

##### The Code/Explanation of Method [29]

  1) a) We begin by creating a synthetic dataset to classify. We use the "make-moons" package from sklearn to generate spherical clusters of points. We also assign one of two features (arbitrarily named Feature 1 and Feature 2) to every point. The model will then classify based on these features.

```python
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay


# Create synthetic 2D data
X_KNNData, y_KNNData = make_moons(n_samples=300, noise=0.35, random_state=71)

# Create a DataFrame for plotting
df_KNN = pd.DataFrame(X_KNNData, columns=["Feature 1", "Feature 2"])
df_KNN['Target'] = y_KNNData

# Visualize the 2D data
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_KNN, x="Feature 1", y="Feature 2", hue="Target", palette="Set1")
plt.title("2D Classification Data (make_moons)")
plt.grid(True)
plt.show()
```

  1) b) We finish establishing the dataset by normalizing its features. This ensures each distance is weighted equally when the model performs its calculations.

  2) Then, we split the data into training and testing sections. A 70/30 split between training and testing respectively is used for this example. It is a common split seen in K-Nearest Neighbor models.

```python
# Normalize the features
scaler_KNN = StandardScaler()
X_scaled_KNN = scaler_KNN.fit_transform(X_KNNData) #z-score normalization to ensure each distance weighted equally

# Split into train and test
#70% training, 30% testing
X_train_KNN, X_test_KNN, y_train_KNN, y_test_KNN = train_test_split(
    X_scaled_KNN, y_KNNData, test_size=0.3, random_state=71, stratify=y_KNNData
)
```

  3) Then, we use the KNeighborsClassifier from the SkLearn package to generate and train the model. A parameter for this model is the *K*-value. We begin with $K=5$ for simplicity. This means the nearest 5 data points will be taken into account when classifying the point as having Feature 1 or Feature 2.

  * Within the model creation, unseen calculations are occurring. The model takes a point, calculates the distances between that point and all the other datapoints, retrieves the five closest points, and completes a "voting" procedure where it assigns a Feature to the original datapoint. It proceeds until all data has been accounted for.


```python
#creating the actual model
# Train a k-NN classifier
knn = KNeighborsClassifier(n_neighbors=5) #we start with k=5
knn.fit(X_train_KNN, y_train_KNN) #trains model

# Predict and evaluate
y_pred_KNNMODEL = knn.predict(X_test_KNN)
print(f"Test Accuracy (k=5): {accuracy_score(y_test_KNN, y_pred_KNNMODEL):.2f}")
```

  Note: The accuracy found from the original model is 82%. This is relatively high for a first iteration. However, it is still important to cross-validate the model by testing other *K*-values.

  4) Complete model cross-validation by generating new models for a range of *K*-values and graphing their accuracy against one another.

```python
##CROSS VALIDATION OF MODEL
#ensuring k-value is appropriate to try

# Range of k values to try
k_range_KNNMODEL = range(1, 21)
cv_scores_KNNMODEL = []

# Evaluate each k using 5-fold cross-validation
##calculates accuracy of the model for each k-value
for k in k_range_KNNMODEL:
    knn = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(knn, X_scaled_KNN, y_KNNData, cv=5, scoring='accuracy')
    cv_scores_KNNMODEL.append(scores.mean())

# Plot accuracy vs. k
plt.figure(figsize=(8, 5))
plt.plot(k_range_KNNMODEL, cv_scores_KNNMODEL, marker='o')
plt.title("k-NN Cross-Validation Accuracy vs k")
plt.xlabel("Number of Neighbors: k")
plt.ylabel("Cross-Validated Accuracy")
plt.grid(True)
plt.show()

# Best k
best_k_KNNMODEL = k_range_KNNMODEL[np.argmax(cv_scores_KNNMODEL)]
print(f"Best k from cross-validation: {best_k_KNNMODEL}")
```

  * In this case, the best *K*-value is found to be 7. However, it is clear that the fluctuations in values of K can lead to a wide spread of model accuracy.

  5) Generate the final model with the ideal value of *K*. Predict new values using test data.

```python
# Train final model with best k
best_knn_KNN = KNeighborsClassifier(n_neighbors=best_k_KNNMODEL)
best_knn_KNN.fit(X_train_KNN, y_train_KNN)

# Predict on test data
y_pred_KNNModel = best_knn_KNN.predict(X_test_KNN)
```

  6) A further evaluation of the model is found below. A confusion matrix is used to find the values for precision, recall, F1 score, and support. These measures are explained below.

  * **Precision** - Measures the accuracy of "positive" predictions (How many predicted Feature 1 points actually were Feature 1?)
  * **Recall** - Evaluates model's ability to predict "positives" (Out of all the Feature 1 points, how many did the model accurately identify?)
  * **F1 Score** - The harmonic mean of precision and recall
  * **Support** - The number of occurrences of each label in the data-set

```python
#USING MORE EVALUATION MEASURES [12]

# Confusion Matrix
cm_KNN = confusion_matrix(y_test_KNN, y_pred_KNNModel)
disp_KNN = ConfusionMatrixDisplay(confusion_matrix=cm_KNN, display_labels=["Class 0", "Class 1"])
disp_KNN.plot(cmap="Blues")
plt.title(f"Confusion Matrix (k={best_k_KNNMODEL})")
plt.grid(False)
plt.show()

# Detailed classification report
print("Classification Report:")
print(classification_report(y_test_KNN, y_pred_KNNModel, target_names=["Class 0", "Class 1"]))
```

  We then visualize the datapoints again with the new decision boundary to demonstrate accuracy visually.

```python
#Graphing the points with the decision boundary (using best k-value)

# Create mesh grid
x_min_KNN, x_max_KNN = X_scaled_KNN[:, 0].min() - 1, X_scaled_KNN[:, 0].max() + 1
y_min_KNN, y_max_KNN = X_scaled_KNN[:, 1].min() - 1, X_scaled_KNN[:, 1].max() + 1

xx_KNN, yy_KNN = np.meshgrid(
    np.linspace(x_min_KNN, x_max_KNN, 300),
    np.linspace(y_min_KNN, y_max_KNN, 300)
)
# Predict on mesh grid with best k
Z_KNN = best_knn_KNN.predict(np.c_[xx_KNN.ravel(), yy_KNN.ravel()])
Z_KNN = Z_KNN.reshape(xx_KNN.shape)

# Plot decision boundary
plt.figure(figsize=(8, 6))
plt.contourf(xx_KNN, yy_KNN, Z_KNN, cmap=plt.cm.coolwarm, alpha=0.3)
sns.scatterplot(x=X_scaled_KNN[:, 0], y=X_scaled_KNN[:, 1], hue=y_KNNData, palette="Set1", edgecolor='k')
plt.title(f"Decision Boundary with Best k = {best_k_KNNMODEL}")
plt.xlabel("Feature 1 (scaled)")
plt.ylabel("Feature 2 (scaled)")
plt.grid(True)
plt.show()
```

# Clustering
## K-Means Clustering

*This section utilizes information from [30,31]*

### Background Information:

$\textbf{What is it?}$

K-Means Clustering is an **exclusive clustering method** used to classify data points into groups based on their characteristics [31]. "Exclusive" denotes how data points can only belong to one cluster. This technique is used on unlabeled data points, where there is no predefined classification structure [30].

<bk>

$\textbf{Why is it useful?}$
* Data Segmentation [30]
  * Businesses can use K-Means Clustering to group customers by shopping habits or website interaction.
* Image Compression [31]
  * K-Means Clustering algorithms are used to cluster and average pixels to reduce image size.
* Anomaly Detection [31]
  * These algorithms are especially proficient at identifying points that do not lie within given clusters.

<bk>

$\textbf{What are the challenges?}$
* Selecting the appropriate **number of clusters to identify** (the value of *k*) is a challenge when the data-set is complex [31]. However, this is mitigated through the "Elbow Method," which chooses *k* based on where change in "Within Cluster Sum of Squares" values level off ("Within Cluster Sum of Squares" is an accuracy measure).

* When **non-spherical clusters or outliers** are present, K-Means Clustering algorithms struggle because spherical data and no outliers are assumptions of the model [30].

* K-Means Clustering models are extremely **sensitive to initial centroid positions**, which can lead to inconsistent final results [30]. The K-Means++ Algorithm seeks to limit these errors by picking initial centroids iteratively based on the positions of previous centroids [31].

### The Code/Explanation of Method [30]:

In the K-Means Clustering model, the center of each cluster is known as a **centroid.** The following example uses the SkLearn package to create a simple K-Means Clustering based on randomly generated data. We begin by importing the relevant packages.

```python
#import relevant packages
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

  1) The dataset is generated using the "make_blobs" package, which generates points in pre-determined clusters. Four clusters are used in this model. These points are plotted for visual display.

```python
#generate random dataset (points in pre-determined clusters)
X_kmeans,y_kmeans = make_blobs(n_samples = 500,n_features = 2,centers = 4,random_state = 40)

fig = plt.figure(0)
plt.grid(True)
plt.scatter(X_kmeans[:,0],X_kmeans[:,1])
plt.show()
```

  2) a) The value of *k* is assigned. *k* denotes the number of clusters the model will seek to classify (In this case, we set $k=4$ to align with the data generated. However, in other data-sets, multiple *k*-values would need to be tested.).
  
  2) b) The centroid positions are randomly generated initially.

```python
#initialize centroids randomly
k_kmeans = 4

clusters_kmeans = {}
np.random.seed(35)

for idx in range(k_kmeans):
    center_kmeans = 2*(2*np.random.random((X_kmeans.shape[1],))-1)
    points_kmeans = []
    cluster_kmeans = {
        'center' : center_kmeans,
        'points' : []
    }

    clusters_kmeans[idx] = cluster_kmeans

clusters_kmeans
```

  2) c) We plot these points for reference.

```python
#plot the data points and the initialized centroids
plt.scatter(X_kmeans[:,0],X_kmeans[:,1])
plt.grid(True)
for i in clusters_kmeans:
    center_kmeans = clusters_kmeans[i]['center']
    plt.scatter(center_kmeans[0],center_kmeans[1],marker = '*',c = 'red')
plt.show()
```

  3) The measure of distance is defined for this model. Euclidean distance is one of the most common "distances" to utilize, so it is implemented here.

```python
#define euclidean distance, the measure of correlation used
def euclidean_distance_kmeans(p1,p2):
    return np.sqrt(np.sum((p1-p2)**2))
```

  4) Now, the model is created. First, each datapoint is assigned to a cluster (based on its distance to the nearest centroid). Then, the positions of the centroids is re-evaluated by averaging the positions of all the datapoints within the cluster. This process is repeated until the positions of the centroids stabilizes (or does not move within a tolerance distance).

```python
#the algorithm itself
#assigning data points to clusters
def assign_clusters_kmeans(X_kmeans, clusters_kmeans):
    for idx in range(X_kmeans.shape[0]):
        dist_kmeans = []

        curr_x_kmeans = X_kmeans[idx]

        for i in range(k_kmeans):
            dis_kmeans = euclidean_distance_kmeans(curr_x_kmeans,clusters_kmeans[i]['center'])
            dist_kmeans.append(dis_kmeans)
        curr_cluster_kmeans = np.argmin(dist_kmeans)
        clusters_kmeans[curr_cluster_kmeans]['points'].append(curr_x_kmeans)
    return clusters_kmeans

#updating centroids and data points based on new clusters
def update_clusters_kmeans(X_kmeans, clusters_kmeans):
    for i in range(k_kmeans):
        points_kmeans = np.array(clusters_kmeans[i]['points'])
        if points_kmeans.shape[0] > 0:
            new_center_kmeans = points_kmeans.mean(axis =0)
            clusters_kmeans[i]['center'] = new_center_kmeans

            clusters_kmeans[i]['points'] = []
    return clusters_kmeans
```

  5) The model is then asked to predict clusters based on new test data.

```python
#prediction of new centroids
def pred_cluster_kmeans(X_kmeans, clusters_kmeans):
    pred_kmeans = []
    for i in range(X_kmeans.shape[0]):
        dist_kmeans = []
        for j in range(k_kmeans):
            dist_kmeans.append(euclidean_distance_kmeans(X_kmeans[i],clusters_kmeans[j]['center']))
        pred_kmeans.append(np.argmin(dist_kmeans))
    return pred_kmeans
```

Below, we actually run the algorithms established above on our given dataset. The results are plotted.

```python
#running the algorithm

clusters_kmeans = assign_clusters_kmeans(X_kmeans,clusters_kmeans)
clusters_kmeans = update_clusters_kmeans(X_kmeans,clusters_kmeans)
pred_kmeans = pred_cluster_kmeans(X_kmeans,clusters_kmeans)

plt.scatter(X_kmeans[:,0],X_kmeans[:,1],c = pred_kmeans)
for i in clusters_kmeans:
    center_kmeans = clusters_kmeans[i]['center']
    plt.scatter(center_kmeans[0],center_kmeans[1],marker = '^',c = 'red')
plt.show()
```

Observe that these clusters do not align directly with the "obvious" ones generated by the dataset. Like all models, K-Means Clustering is also prone to errors.
